```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import json
from scipy import stats as st
import os
import csv
import glob
```

```python
def analysis(df):
    df_columns = list(df.columns)
    unique_url = len(df)
    unique_fqdn = len(set(df['FQDN']))

    live_ham, live_spam, live_smish = 0,0,0
    if 'ham' in df_columns:
        live_ham = len(df[(df['Status Code']==200) & (df['ham']==1)])
    if 'spam' in df_columns:
        live_spam = len(df[(df['Status Code']==200) & (df['spam']==1)])
    if 'smish' in df_columns:
        live_smish = len(df[(df['Status Code']==200) & (df['smish']==1)])
    if 'smishing' in df_columns:
        live_smish = len(df[(df['Status Code']==200) & (df['smishing']==1)])
    total_live = live_ham + live_spam + live_smish

    parked_ham, parked_spam, parked_smish = 0,0,0
    if 'ham' in df_columns:
        parked_ham = len(df[(df['Parked']==1) & (df['ham']==1)])
    if 'spam' in df_columns:
        parked_spam = len(df[(df['Parked']==1) & (df['spam']==1)])
    if 'smish' in df_columns:
        parked_smish = len(df[(df['Parked']==1) & (df['smish']==1)])
    if 'smishing' in df_columns:
        parked_smish = len(df[(df['Parked']==1) & (df['smishing']==1)])
    total_parked = parked_ham + parked_spam + parked_smish

    # print("Unique URL   Unique FQDN    total_live_URL   ham_live    spam_live   smish_live  totla_parked    ham_parked  spam_parked     smish_parked")
    # print(unique_url,unique_fqdn,total_live,live_ham,live_spam,live_smish,total_parked,parked_ham,parked_spam,parked_smish)

    
    # The total_live_URL column needs a bit more space, so it's set to 18.
    header_format = (
        f"{'Unique URL':<15}"
        f"{'Unique FQDN':<15}"
        f"{'total_live_URL':<18}" 
        f"{'ham_live':<12}"
        f"{'spam_live':<12}"
        f"{'smish_live':<12}"
        f"{'total_parked':<15}"
        f"{'ham_parked':<12}"
        f"{'spam_parked':<12}"
        f"{'smish_parked':<12}"
    )

    # --- Define the Data Format ---
    data_format = (
        f"{unique_url:^15}"
        f"{unique_fqdn:^15}"
        f"{total_live:^18}"
        f"{live_ham:^12}"
        f"{live_spam:^12}"
        f"{live_smish:^12}"
        f"{total_parked:^15}"
        f"{parked_ham:^12}"
        f"{parked_spam:^12}"
        f"{parked_smish:^12}"
    )

    # --- Print the Table ---
    print(header_format)
    print("-" * len(header_format)) 
    print(data_format)
```

```python
##This will only work for the labeld dat, so ignore it for the unlabel ones. 


folder_path = '../URL Data/'
file_pattern = 'Websites Analysis.csv'
search_path = os.path.join(folder_path, '*' + file_pattern)
all_files = glob.glob(search_path)
for file in all_files:
    df = pd.read_csv(file)
    print('analysis for the dataset:', file[len(folder_path):-4])
    analysis(df)
```
